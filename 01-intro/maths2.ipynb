```python
def vec_dot_prod(vec1, vec2):
    assert vec1.shape[0]==vec2.shape[0]
    n = vec1.shape[0]
    sum = 0
    for i in range(n):
        sum += vec1[i]*vec2[i]
    print(sum)
    return sum
```

```python
vec_dot_prod(arr, v)
```

```python
U = np.random.randint(1,20,(3,4))
U
```

```python
v = np.hstack([v, 5])
v
```

```python
def matrix_vec_mul(mat, vec):
    assert mat.shape[1] == vec.shape[0]
    m = mat.shape[0]
    result = np.zeros(m)
    for j in range(m):
        result[j] = vec_dot_prod(mat[j], vec)
    return result
```

```python
matrix_vec_mul(U, v)
```

```python
U.dot(v)
```

```python
V = np.random.randint(2,25, (4,3))
V
```

```python
def mat_mat_mul(mat1, mat2):
    assert mat1.shape[1] == mat2.shape[0]
    r = mat1.shape[0]
    c = mat2.shape[1]
    array = np.zeros((r,c))
    for i in range(c):
        vi = mat2[:, i]
        Uvi = matrix_vec_mul(mat1, vi)
        array[:, i] = Uvi
    return array
```

```python
mat_mat_mul(U, V)
```
